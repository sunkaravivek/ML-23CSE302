```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
data = pd.DataFrame({
    'Weight': [79,69,73,95,82,55,69,71,64,69],
    'Height': [1.80,1.68,1.82,1.70,1.87,1.55,1.50,1.78,1.67,1.64],
    'Age': [35,39,25,60,27,18,89,42,16,52],
    'Gender': ['Male','Male','Male','Male','Male',
               'Female','Female','Female','Female','Female']
})
```

```python
data.head()
```

```python
data.shape
```

```python
data.isnull().sum()
```

```python
data['Gender'] = data['Gender'].map({
    'Male':1,
    'Female':0
})
```

```python
data.head()
```

```python
X = data.iloc[:,1:].values
```

```python
y = data.iloc[:,0].values
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
   random_state=1
)
```

```python
model = LinearRegression()

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print("Predicted Weight")
print(y_pred)

print("\nActual Weight")
print(y_test)
```

```python
new_prediction = model.predict([[1.75,30,1]])

print("Predicted Weight:",new_prediction[0],"kg")
```

```python
plt.scatter(data['Height'],
            data['Weight'])

plt.xlabel("Height (m)")
plt.ylabel("Weight (kg)")
plt.title("Height vs Weight")
plt.grid(True)
plt.show()
```

```python
print("Training Score :",model.score(X_train,y_train))

print("Testing Score :",model.score(X_test,y_test))
```
