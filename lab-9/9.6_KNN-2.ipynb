```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

# Dataset
data = {
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class': ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']
}

df = pd.DataFrame(data)

# Features
X = df[['Brightness', 'Saturation']]

# Target
y = df['Class']

# Create KNN model
model = KNeighborsClassifier(n_neighbors=3)

# Train the model
model.fit(X, y)

# New entry
new_entry = [[50, 40]]

# Predict
prediction = model.predict(new_entry)

print("New Entry:", new_entry)
print("Predicted Class:", prediction[0])
```

```python
distances, indices = model.kneighbors(new_entry)

print("Distances:")
print(distances)

print("Nearest Classes:")
print(y.iloc[indices[0]].values)
```
